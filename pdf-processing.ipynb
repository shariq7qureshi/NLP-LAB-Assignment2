```python
import PyPDF2
import re
import string
import pandas as pd
import nltk
import plotly.express as px

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
pdf_path = "book.pdf"

text = ""

with open(pdf_path, "rb") as file:
    reader = PyPDF2.PdfReader(file)

    total_pages = len(reader.pages)

    pages_to_read = min(100, total_pages)

    for page_num in range(pages_to_read):
        page = reader.pages[page_num]
        text += page.extract_text()

print("Total Pages in PDF:", total_pages)
print("Pages Extracted:", pages_to_read)
```

```python
print(text[:1000])
```

```python
lower_text = text.lower()

print(lower_text[:500])
```

```python
no_numbers = re.sub(r'\d+', '', lower_text)

print(no_numbers[:500])
```

```python
no_special = re.sub(r'[^a-zA-Z\s]', '', no_numbers)

print(no_special[:500])
```

```python
clean_text = re.sub(r'\s+', ' ', no_special).strip()

print(clean_text[:500])
```

```python
clean_text = clean_text.translate(
    str.maketrans('', '', string.punctuation)
)

print(clean_text[:500])
```

```python
tokens = word_tokenize(clean_text)

print(tokens[:50])
```

```python
stop_words = set(stopwords.words('english'))

valid_words = [
    word for word in tokens
    if word not in stop_words
]

stopword_count = len(tokens) - len(valid_words)

print("Stop Words Count:", stopword_count)
print("Valid Words Count:", len(valid_words))
```

```python
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in valid_words
]

print(stemmed_words[:50])
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in valid_words
]

print(lemmatized_words[:50])
```

```python
sample_words = list(set(lemmatized_words[:20]))

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(
    pd.DataFrame(sample_words)
)

onehot_df = pd.DataFrame(
    encoded,
    columns=encoder.categories_[0]
)

print(onehot_df)
```

```python
final_text = " ".join(lemmatized_words)

vectorizer = TfidfVectorizer(max_features=20)

tfidf_matrix = vectorizer.fit_transform([final_text])

feature_names = vectorizer.get_feature_names_out()

tfidf_scores = tfidf_matrix.toarray()[0]

tfidf_df = pd.DataFrame({
    "Word": feature_names,
    "TF-IDF Score": tfidf_scores
})

print(tfidf_df)
```

```python
fig = px.scatter(
    tfidf_df,
    x="Word",
    y="TF-IDF Score",
    title="TF-IDF Scatter Plot",
    size="TF-IDF Score"
)

fig.show()
```

```python
fig = px.bar(
    tfidf_df,
    x="Word",
    y="TF-IDF Score",
    title="Top TF-IDF Words"
)

fig.show()
```

```python
fig = px.line(
    tfidf_df,
    x="Word",
    y="TF-IDF Score",
    title="TF-IDF Line Graph"
)

fig.show()
```
